# Subproblem (a)

```python
from sympy import *
```

```python
x = Symbol('x')
g = 4 * x**3 + 3 * x**2 - 6 * x - 1
g
```

```python
for i in range(-2, 3):
    gx = g.subs(x, i)
    print(f"g({i:2}) = {gx}")
```

```python
plot(g, xlim=(-2,2), ylim=(-4,4))
```

# Subproblem (b)

By inspecting the values above: $x=1$ is a root! And this means $g(x)$ can be written in the form of $g(x) = (x-1) (a x^2 + b x + c)$.

```python
a,b,c = symbols('a,b,c')
expand((x-1)*(a*x**2 + b*x + c))
```

which we compare to

```python
g
```

and deduce that:
- $a = 4$ (from the $x^3$ term)
- $b - a = 3$, substituting $a = 4$ gives $b = 7$ (from the $x^2$ term)
- $c - b = -6$, substituting $b = 7$ yields $c = 1$ (from the $x$ term)
- $c = 1$ (from the constant term) is the verification that we didn't mess up in the calculations above.

```python
a = 4
b = 7
c = 1
g1 = a * x**2 + b * x + c
expand((x-1) * g1) == g # another check
```

```python
from IPython.display import display, Math

display((-b + sqrt(b**2 - 4*a*c))/2*a)
display((-b + sqrt(b**2 - 4*a*c))/2*a)
```

# Subproblem (c)

```python
k = symbols('k')
f = 4 * x**3 + 3 * x**2 - 6 * x + k
f
```

$f$ must have **at least one** real root because $f(-\infty) = -\infty$ and $f(+\infty) = + \infty$, so $f$ must cross the $x$ axis.

```python
df = f.diff(x)
df
```

```python
d1, d2 = solve(df, x)
display(d1)
display(d2)
```

```python
display(f.subs(x, d1))
display(f.subs(x, d2))
```

$f$ has a local maxima $f(x) = k + 5$ at $x = -1$ and a local minima $f(x) = k - 7/4$ at $x = 1/2$.

- If $k + 5 < 0 \iff k < -5$ (*e.g.* $k = -6$) then the root in $(-\infty, -1)$ and the root in $(-1, 1/2)$ disappears.

```python
plot(f.subs(k, -6), xlim=(-2, 2), ylim=(-10, 10))
```

- If $k + 5 \ge 0 \land k - 7/4 \le 0 \iff -5 \le k \le 7/4$ (*e.g.* $k = 2$) then there are 3 roots.

```python
plot(f.subs(k, 0), xlim=(-2, 2), ylim=(-10, 10))
```

- If $k - 7/4 > 0 \iff k > 7/4$ (*e.g.* $k = 2$) then the root in $(-1, 1/2)$ and the root in $(1/2, +\infty)$ disappears.

```python
plot(f.subs(k, 2), xlim=(-2, 2), ylim=(-10, 10))
```
