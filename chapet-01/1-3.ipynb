In the above expression, if $x > 0$ then all terms are positive, with the exception of $4 (x - 0.375)^2$, which is non-negative (*i.e.* can be zero), meaning that **the sum of all terms is positive**.

This means that the derivative
- has no local minima or maxima in the $(0, +\infty)$ interval. Together with:
- $f'(0) = 0$ (see above)
implies that there can be at most one root in the $(0, +\infty)$ interval, and considering
- $f(0) = -2 < 0$ (see below),
- $f(+\infty) = +\infty$
there is exactly one root in $(0, +\infty)$.
